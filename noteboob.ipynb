```python
import seaborn as sb
import polars as pl
```

```python
df = pl.scan_csv("penalties_py_0.txt", separator="\t")
```

```python
pos_df = df.filter(pl.col("Reaction").str.ends_with("_pos")).select(pl.col("Reaction").str.replace("_pos", ""), pl.selectors.numeric())
neg_df = df.filter(pl.col("Reaction").str.ends_with("_neg")).select(pl.col("Reaction").str.replace("_neg", ""), pl.selectors.numeric())
```

```python
pos_df.head().collect()
```

```python
combined_df = pos_df.join(neg_df, on="Reaction", how = "inner", suffix="_neg")
```

```python
combined_df.head().collect()
```

```python
sample = "Ob-DHA-e_S154_L007_R1_001"
combined_df.select(pl.col(sample).eq(pl.col(sample + "_neg")).alias("equal_pn").all()).collect()
```

Ok, so yeah the pos and neg are always equal to each other.

```python
df_py = pl.scan_csv("penalties_py_0.txt", separator="\t")
```

```python
df_rs = pl.scan_csv("crates/penalties.csv")
df_rs_pos = df_rs.select(pl.col("_reaction_name").alias("Reaction") + "_pos", pl.selectors.numeric())
df_rs_neg = df_rs.select(pl.col("_reaction_name").alias("Reaction") + "_neg", pl.selectors.numeric())
df_rs = df_rs_pos.collect().extend(df_rs_neg.collect()).lazy()
```

```python
sample = "Ob-DHA-e_S154_L007_R1_001"
df_py.join(df_rs, on="Reaction", how="inner", suffix="_rs").filter(~pl.col(sample).eq(pl.col(sample + "_rs")).alias("not equal")).select(pl.col("Reaction"), pl.col(sample), pl.col(sample + "_rs")).collect()
```

```python
df_rs = pl.scan_csv("crates/penalties_iso.csv")
df_rs_pos = df_rs.select(pl.col("_reaction_name").alias("Reaction") + "_pos", pl.selectors.numeric())
df_rs_neg = df_rs.select(pl.col("_reaction_name").alias("Reaction") + "_neg", pl.selectors.numeric())
df_rs = df_rs_pos.collect().extend(df_rs_neg.collect()).lazy()
sample = "Ob-DHA-e_S154_L007_R1_001"
df_py.join(df_rs, on="Reaction", how="inner", suffix="_rs").filter(~pl.col(sample).eq(pl.col(sample + "_rs")).alias("not equal")).select(pl.col("Reaction"), pl.col(sample), pl.col(sample + "_rs")).collect()
```
